task 1.
Create a line chart using Matplotlib to show the number of daily active users for a music streaming app (like Spotify) over 12 months, and use plt.annotate() to label the month with the highest user count.


```python
import matplotlib.pyplot as plt
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
```

```python
users = [1000, 1200, 1500, 1300, 1800, 2000,
         2200, 2100, 2500, 2800, 2600, 3200]
```

```python
plt.plot(months, users, marker='o', color='green')

max_users = max(users)
max_month = months[users.index(max_users)]

plt.annotate("Highest Users",
             xy=(max_month, max_users),
             xytext=(max_month, max_users - 500),
             arrowprops=dict(arrowstyle='->', color='red'))

plt.xlabel("Months")
plt.ylabel("Daily Active Users")
plt.title("Spotify Monthly Active Users")

plt.show()
```

task 2.
Plot a bar chart showing the number of orders per month for a food delivery app (like Zomato), and add an arrow annotation using plt.annotate() to highlight the month with the lowest orders.<br><br><em><strong>Hint:</strong> Use the 'arrowprops' parameter to draw the arrow pointing to the bar.</em>


```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
```

```python
orders = [120, 150, 100, 180, 200, 170,
          220, 190, 160, 210, 240, 230]
```

```python
plt.bar(months, orders, color='orange')

min_orders = min(orders)
min_month = months[orders.index(min_orders)]

plt.annotate("Lowest Orders",
             xy=(min_month, min_orders),
             xytext=(min_month, min_orders + 40),
             arrowprops=dict(arrowstyle='->', color='red'),
             color='red')

plt.xlabel("Months")
plt.ylabel("Number of Orders")
plt.title("Monthly Zomato Orders")

plt.show()
```

task 3.
Create a pie chart representing the percentage of different payment methods (UPI, Card, Wallet, Cash) used on an e-commerce app (like Flipkart), and add a legend positioned at the lower right corner of the plot.


```python
methods = ["UPI", "Card", "Wallet", "Cash"]
payments = [50, 25, 15, 10]
```

```python
plt.pie(payments, labels=methods, autopct='%1.1f%%')

plt.legend(loc='lower right')

plt.title("Flipkart Payment Methods")

plt.show()
```

task 4.
Given a scatter plot of movie ratings vs. number of reviews for movies on a streaming platform (like BookMyShow), use plt.annotate() to add a text box highlighting the movie with the highest rating.<br><br><em><strong>Hint:</strong> Set the 'bbox' parameter in plt.annotate() to create a visible text box.</em>


```python
movies = ["Movie A", "Movie B", "Movie C", "Movie D",
          "Movie E", "Movie F"]
```

```python
ratings = [7.5, 8.2, 9.1, 6.8, 8.5, 7.9]
reviews = [1200, 2000, 3500, 900, 2800, 1500]

```

```python
plt.scatter(ratings, reviews, color='blue')

max_rating = max(ratings)
index = ratings.index(max_rating)

plt.annotate(
    movies[index],
    xy=(ratings[index], reviews[index]),
    xytext=(ratings[index] - 1, reviews[index] + 300),
    bbox=dict(boxstyle='round', facecolor='yellow', edgecolor='black'),
    arrowprops=dict(arrowstyle='->', color='red')
)

plt.xlabel("Movie Rating")
plt.ylabel("Number of Reviews")
plt.title("Movie Ratings vs Reviews")

plt.show()
```
