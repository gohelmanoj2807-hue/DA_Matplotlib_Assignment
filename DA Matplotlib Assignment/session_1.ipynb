task 3.
Write a Python script that creates a Matplotlib figure with two Axes: one showing your daily Spotify listening hours for a week and the other showing your daily YouTube viewing hours for the same week. Display both plots in a single figure.


```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
```

```python
spotify = [2, 3, 1, 2, 4, 3, 2]
```

```python
youtube = [1, 2, 3, 2, 3, 4, 5]
```

```python
fig, ax = plt.subplots(1, 2)
```

```python
ax[0].plot(days, spotify, marker='o') 
ax[0].set_title("Spotify Listening Hours") 
ax[0].set_xlabel("Days")
ax[0].set_ylabel("Hours")
```

```python
ax[1].plot(days, youtube, marker='o') 
ax[1].set_title("YouTube Viewing Hours")
ax[1].set_xlabel("Days")
ax[1].set_ylabel("Hours")
```

```python
plt.tight_layout()
```

```python
plt.show()
```
