task 1.
Use plt.hist() to plot a histogram of delivery times (in minutes) for 30 Swiggy orders: [32, 28, 45, 30, 36, 40, 29, 34, 38, 31, 27, 41, 33, 35, 39, 37, 44, 30, 29, 32, 36, 28, 43, 31, 35, 40, 38, 33, 30, 42]. Show the plot window.


```python
import matplotlib.pyplot as plt
```

```python
delivery_times = [32, 28, 45, 30, 36, 40, 29, 34, 38, 31,
                  27, 41, 33, 35, 39, 37, 44, 30, 29, 32,
                  36, 28, 43, 31, 35, 40, 38, 33, 30, 42]
```

```python
plt.hist(delivery_times, bins=5, color='orange', edgecolor='black')

plt.xlabel("Delivery Time (minutes)")
plt.ylabel("Number of Orders")
plt.title("Swiggy Delivery Time Distribution")

plt.show()
```

task 3.
Given an array of daily step counts from a fitness tracker app for one month, plot two histograms using plt.hist(): one showing frequency (default), and one showing density=True. Compare the y-axis labels and explain the difference.


```python
steps = [4500, 6000, 7500, 8000, 9000, 5000, 6500,
         7000, 8500, 10000, 5500, 7200, 6800, 9200,
         7800, 8300, 6100, 7400, 9600, 8800, 5300,
         6900, 8100, 7700, 9400, 6200, 8600, 7100,
         9800, 6700]
```

```python
plt.hist(steps, bins=5, color='blue', edgecolor='black')
plt.xlabel("Daily Steps")
plt.ylabel("Frequency (Number of Days)")
plt.title("Daily Step Counts - Frequency")
plt.show()
```

```python
plt.hist(steps, bins=5, density=True, color='orange', edgecolor='black')
plt.xlabel("Daily Steps")
plt.ylabel("Density")
plt.title("Daily Step Counts - Density")
plt.show()
```

task 4.
Simulate the distribution of song durations (in seconds) from your Spotify playlist by generating 50 random numbers between 120 and 300. Plot a histogram using plt.hist() and set the color to 'purple' and add the title 'Spotify Song Duration Distribution'.


```python
import random
```

```python
durations = []

for i in range(50):
    durations.append(random.randint(120, 300))
```

```python
plt.hist(durations, bins=10, color='purple', edgecolor='black')
plt.xlabel("Song Duration (seconds)")
plt.ylabel("Number of Songs")
plt.title("Spotify Song Duration Distribution")

plt.show()
```
