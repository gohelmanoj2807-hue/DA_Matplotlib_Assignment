task 2.
Create a scatter plot showing 'amount spent on Zomato orders' vs 'number of orders placed' for 12 friends, and label each point with the friend's name.<br><br><em><strong>Hint:</strong> Use the 'annotate' function in matplotlib to add labels.</em>


```python
names = ["Amit", "Raj", "Priya", "Neha", "Ravi", "Kiran",
         "Manoj", "Pooja", "Jay", "Riya", "Vijay", "Ankit"]
```

```python
orders = [5, 8, 4, 10, 6, 12, 7, 9, 3, 11, 6, 8]
spending = [1000, 1800, 800, 2500, 1200, 3000,
            1500, 2200, 600, 2800, 1300, 1900]

```

```python
plt.scatter(orders, spending, color='green')

for i in range(len(names)):
    plt.annotate(names[i], (orders[i], spending[i]))

plt.xlabel("Number of Orders")
plt.ylabel("Amount Spent (Rs.)")
plt.title("Zomato Orders vs Spending")

plt.show()
```

task 3.
Add a trend line to your Zomato scatter plot by calculating the best-fit line using numpy's polyfit(), then plot it on top of the scatter plot.<br><br><em><strong>Hint:</strong> Use np.polyfit() to get the slope and intercept, then plot the line with plt.plot().</em>


```python
import numpy as np
```

```python
names = ["Amit", "Raj", "Priya", "Neha", "Ravi", "Kiran",
         "Manoj", "Pooja", "Jay", "Riya", "Vijay", "Ankit"]
```

```python
orders = [5, 8, 4, 10, 6, 12, 7, 9, 3, 11, 6, 8]
spending = [1000, 1800, 800, 2500, 1200, 3000,
            1500, 2200, 600, 2800, 1300, 1900]
```

```python
plt.scatter(orders, spending, color='green', label='Friends')

for i in range(len(names)):
    plt.annotate(names[i], (orders[i], spending[i]))
m, b = np.polyfit(orders, spending, 1)
plt.plot(orders, np.array(orders) * m + b,
         color='red', label='Trend Line')

plt.xlabel("Number of Orders")
plt.ylabel("Amount Spent (Rs.)")
plt.title("Zomato Orders vs Spending with Trend Line")

plt.legend()
plt.show()
```

task 4.
Given a dataset of 'YouTube watch hours' and 'Spotify listening hours' for 15 users, use numpy polyfit to fit a regression line, plot both the scatter and the trend line, and print the equation of the line on the plot.


```python
youtube = [2, 3, 1, 5, 4, 6, 2, 7, 3, 5, 8, 4, 6, 9, 7]
spotify = [3, 4, 2, 6, 5, 7, 3, 8, 4, 6, 9, 5, 7, 10, 8]
```

```python
plt.scatter(youtube, spotify, color='blue', label='Users')

m, b = np.polyfit(youtube, spotify, 1)

plt.plot(youtube, np.array(youtube) * m + b,
         color='red', label='Regression Line')

equation = f"y = {m:.2f}x + {b:.2f}"
plt.text(2, 9, equation, fontsize=10)

plt.xlabel("YouTube Watch Hours")
plt.ylabel("Spotify Listening Hours")
plt.title("YouTube vs Spotify Usage")

plt.legend()
plt.show()

print("Regression Equation:", equation)
```
