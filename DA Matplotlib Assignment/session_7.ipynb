task 1.
Use plt.subplot() to create a figure with 2 rows and 1 column of plots showing your daily step count and water intake (in litres) for the last 7 days. Plot both as line charts, one in each subplot.


```python
import matplotlib.pyplot as plt
```

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
```

```python
steps = [5000, 7000, 6500, 8000, 9000, 10000, 7500]
water = [2, 2.5, 2, 3, 2.5, 3, 2.5]
```

```python
plt.subplot(2, 1, 1)
plt.plot(days, steps, marker='o', color='blue')
plt.title("Daily Step Count")
plt.ylabel("Steps")
```

```python
plt.subplot(2, 1, 2)
plt.plot(days, water, marker='o', color='green')
plt.title("Daily Water Intake")
plt.xlabel("Days")
plt.ylabel("Water (Litres)")

plt.tight_layout()
plt.show()
```

task 2.
Create a 2x2 multi-panel figure using plt.subplots(). In each subplot, display the number of Instagram posts you made in each quarter (Q1, Q2, Q3, Q4) for the past year, using different chart types: line, bar, scatter, and pie.


```python
quarters = ["Q1", "Q2", "Q3", "Q4"]
posts = [20, 35, 25, 40]
```

```python
fig, ax = plt.subplots(2, 2, figsize=(10, 8))
```

```python
ax[0, 0].plot(quarters, posts, marker='o', color='blue')
ax[0, 0].set_title("Line Chart")

ax[0, 1].bar(quarters, posts, color='green')
ax[0, 1].set_title("Bar Chart")

ax[1, 0].scatter(quarters, posts, color='red')
ax[1, 0].set_title("Scatter Plot")
ax[1, 1].pie(posts, labels=quarters, autopct='%1.1f%%')
ax[1, 1].set_title("Pie Chart")

plt.suptitle("Instagram Posts by Quarter")
plt.tight_layout()
plt.show()
```

task 3.
Make a grid of 1 row and 3 columns using plt.subplots(), and plot the number of food orders you placed on Zomato, Swiggy, and Domino's over the last 6 months, one app per subplot. Share the y-axis among all subplots so you can easily compare order counts across apps.<br><br><em><strong>Hint:</strong> Use the 'sharey' parameter in plt.subplots().</em>


```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

zomato = [10, 12, 8, 15, 14, 18]
swiggy = [8, 10, 12, 11, 15, 16]
dominos = [5, 7, 6, 9, 8, 12]

```

```python
fig, ax = plt.subplots(1, 3, figsize=(12, 4), sharey=True)
```

```python
ax[0].plot(months, zomato, marker='o', color='red')
ax[0].set_title("Zomato Orders")
ax[0].set_xlabel("Months")
ax[0].set_ylabel("Number of Orders")

ax[1].plot(months, swiggy, marker='o', color='orange')
ax[1].set_title("Swiggy Orders")
ax[1].set_xlabel("Months")

ax[2].plot(months, dominos, marker='o', color='blue')
ax[2].set_title("Domino's Orders")
ax[2].set_xlabel("Months")

plt.suptitle("Food Orders Comparison")
plt.tight_layout()
plt.show()
```

task 4.
Arrange a 2x2 grid of subplots using plt.subplots() to visualize your monthly mobile data usage (in GB) for four different apps: YouTube, Instagram, Spotify, and WhatsApp. Ensure all subplots share the same x-axis (months) for easy comparison.<br><br><em><strong>Hint:</strong> Use the 'sharex' parameter in plt.subplots().</em>


```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
```

```python
youtube = [5, 7, 6, 8, 10, 12]
instagram = [3, 4, 5, 6, 7, 8]
spotify = [2, 2.5, 3, 3.5, 4, 5]
whatsapp = [1, 1.5, 2, 2, 2.5, 3]
```

```python
fig, ax = plt.subplots(2, 2, figsize=(10, 8), sharex=True)


ax[0, 0].plot(months, youtube, marker='o', color='red')
ax[0, 0].set_title("YouTube Data Usage")
ax[0, 0].set_ylabel("Data (GB)")


ax[0, 1].plot(months, instagram, marker='o', color='purple')
ax[0, 1].set_title("Instagram Data Usage")
ax[0, 1].set_ylabel("Data (GB)")


ax[1, 0].plot(months, spotify, marker='o', color='green')
ax[1, 0].set_title("Spotify Data Usage")
ax[1, 0].set_xlabel("Months")
ax[1, 0].set_ylabel("Data (GB)")


ax[1, 1].plot(months, whatsapp, marker='o', color='blue')
ax[1, 1].set_title("WhatsApp Data Usage")
ax[1, 1].set_xlabel("Months")
ax[1, 1].set_ylabel("Data (GB)")

plt.suptitle("Monthly Mobile Data Usage by App")
plt.tight_layout()
plt.show()
```
